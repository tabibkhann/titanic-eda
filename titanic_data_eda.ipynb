## **Further Analysis**

```python
df.tail(20)
```

Extracting the salution of each person and giving the count

```python
df["Title"] = df["Name"].str.extract(' ([A-Za-z]+)\.')
```

```python
df['Title'].unique()
```

```python
df["Title"].value_counts()
```

```python
def replace_title(row):
    title = row["Title"]
    relevant_titles = ['Mr', 'Mrs', 'Miss', 'Master']
    if title in relevant_titles:
        return title
    else:
        return "Others"
```

```python
df['Title']=df.apply(replace_title,axis=1)
```

```python
df
```
